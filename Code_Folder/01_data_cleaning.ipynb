# Cleaning

```python
#Filter to non-null etiology
df_clean = df[df['Etiology'].notna()].copy()
```

```python
df_clean.Etiology.isna().sum()
```

```python
#Filter to single-etiology outbreaks
df_clean['etiology_list'] = df_clean['Etiology'].str.split(';')
df_clean['etiology_status_list'] = df_clean['Etiology Status'].str.split(';')
```

```python
df_clean = df_clean[(df_clean['etiology_list'].str.len() == 1) & (df_clean['etiology_status_list'].str.len() == 1) & (df_clean['etiology_status_list'].str[0] == 'Confirmed')].copy()
```

```python
df_clean
```

```python
#Convert the single-item list back to a string
df_clean['Etiology'] = df_clean['etiology_list'].str[0]
```

```python
def consolidate_etiology(x):
    if pd.isna(x):
        return x
    if 'Norovirus' in x:
        return 'Norovirus'
    if 'Escherichia coli' in x:
        return 'STEC'
    if 'Salmonella' in x:
        return 'Salmonella'
    if 'Campylobacter' in x:
        return 'Campylobacter'
    if 'Shigella' in x:
        return 'Shigella'
    if 'Clostridium' in x:
        return 'Clostridium'
    if 'Cryptosporidium' in x:
            return 'Cryptosporidium'
    return x

df_clean['Etiology_clean'] = df_clean['Etiology'].apply(consolidate_etiology)
```

```python
top_n = df_clean['Etiology_clean'].value_counts().nlargest(10).index
df_clean['Etiology_top'] = df_clean['Etiology_clean'].where(
    df_clean['Etiology_clean'].isin(top_n), 'Other'
)
```

```python
df_clean['Etiology_top'].value_counts()
```

```python
df_clean = df_clean.drop(columns=['etiology_list', 'etiology_status_list'])
```

```python
df_clean 
```

```python
df_clean["Primary Mode"] = df_clean["Primary Mode"].replace(
    "Environmental contamination other than food/water",
    "Environmental"
)
```

```python
def group_setting(x):
    if pd.isna(x):
        return 'Unknown'
    elif x == 'Long-term care/nursing home/assisted living facility':
        return 'Long-term care'
    elif x == 'Private home/residence':
        return 'Private residence'
    elif x.startswith('Restaurant') or x in ['Caterer', 'Banquet Facility (food prepared and served on-site)']:
        return 'Restaurant'
    elif x in ['School/college/university', 'Child daycare/preschool']:
        return 'School'
    elif x == 'Hospital':
        return 'Hospital'
    elif x == 'Hotel/motel':
        return 'Hotel'
    elif x in ['Community/municipality', 'Religious facility']:
        return 'Community'
    elif x == 'Grocery store/bakery/deli/convenience store':
        return 'Grocery'
    elif x == 'Unknown':
        return 'Unknown'
    else:
        return 'Other'
```

```python
df_clean['Setting_group'] = df_clean['Setting'].apply(group_setting)
df_clean['Setting_group'].value_counts()
```

```python
df_clean = df_clean.reset_index(drop=True)
```

```python
df_clean.to_csv("NORS_cleaned.csv", index=False)
```
